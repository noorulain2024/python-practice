# splitting a string into multiple columns

```python
import pandas as pd
df = pd.DataFrame({'name': ['noor ulain', 'ali raza', 'ahmed ali'],
                   'location':['lahore pakistan', 'grw pakistan', 'multan pakistan']})
df
```

```python
df.name.str.split(' ').head()
```

```python
df[["first_name","last_name"]] = df.name.str.split(' ',expand=True) 
df[["first","last"]] = df.location.str.split(' ',expand=True)
df
```

```python
df= df[['first_name','last_name', 'first', 'last']]
df
```
