```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
kashti = sns.load_dataset("titanic")
kashti
```

```python
kashti.isnull().sum()
```

```python
sns.countplot(x="survived", data=kashti)
```

```python
sns.countplot(x="survived", hue= "sex", data= kashti)
```

```python
plt.scatter(x="age", y= "fare", data= kashti)
```

```python
sns.scatterplot(x="age", y="fare", hue="pclass",data=kashti)
```

```python
sns.barplot(x="sex", y="fare", hue="pclass",data=kashti)
```

```python
sns.pairplot(kashti)
```

```python
kashti.hist(bins=10 , figsize=(9,7), grid=False)
```

```python
sns.boxplot(data=kashti, x="sex", y="fare", hue="class")
```

```python
sns.boxenplot(data=kashti, x="sex", y="fare", hue="class")
```

```python
g=sns.FacetGrid(kashti, hue="survived", col="who",margin_titles=True)
g=g.map(plt.scatter, "age","fare" , edgecolor="w").add_legend()
```

```python
sns.stripplot(data=kashti, x="sex", y="fare", hue="class")
```

```python
corr = g.corr()
plt.figure(figsize=(10,10))
sns.heatmap(corr)
plt.title("heat map of g data")
```

```python
sns.violinplot(data=kashti, x="sex", y="fare", hue="class")
```
